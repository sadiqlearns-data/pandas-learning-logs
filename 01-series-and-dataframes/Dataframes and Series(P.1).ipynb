CREATING A DATAFRAME BY HAND

```python
productsdata = {
    #Columns(Key Values)
    "product": ["Keyboard", "Notebook", "Headphones", "Pen"],#Rows aligned with columns(Values)
    "category": ["Electronics", "Stationery", "Electronics", "Stationery"],
    "stock": [15, 0, 8, 42],
    "unit_price": [450, 35, 750, 15],
}
#This is an example data of products in a market
productsdata
```

```python
productsdataframe=pd.DataFrame(productsdata)#converting into a dataframe
productsdataframe
```

```python
productsdataframe["category"]#Accesing a single column 
```

```python
productsdataframe[['product','stock']] #Accesing multiple columns
```

```python
productsdataframe.columns #accesing to all columns(its useful if there's many columns)
```
